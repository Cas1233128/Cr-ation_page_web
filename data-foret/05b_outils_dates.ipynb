## Aide-mémoire du module 5b

| Je veux… | J'écris |
|---|---|
| Convertir en dates | `pd.to_datetime(df["date"])` |
| Comparer à une date | `df["date"] >= "2026-02-03"` |
| Une durée | `pd.Timedelta(days=1)`, `pd.Timedelta(minutes=30)` |
| Durée en jours / en minutes | `.dt.days`, `.dt.total_seconds() / 60` |
| Heure, jour, mois | `.dt.hour`, `.dt.day`, `.dt.month` |
| Ramener à minuit | `.dt.normalize()` |
| Écart avec la ligne précédente, par groupe | `df.groupby(["station", "espece"])["date"].diff()` |
| Plusieurs résultats nommés par groupe | `.agg(premiere=("date", "min"), derniere=("date", "max"))` |
