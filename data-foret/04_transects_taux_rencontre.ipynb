**b.** Lance le graphique ci-dessous (fourni), puis réponds par écrit :

1. Que montre le graphique entre pression humaine et crottes d'éléphant ?
2. Le coefficient de Spearman est négatif. Peut-on conclure que la chasse **fait fuir** les éléphants ? Cite au moins une autre explication possible.
3. Pourquoi utilise-t-on Spearman ici plutôt que la corrélation de Pearson ? (Indice : regarde la forme du nuage et les nombreux zéros.)

```python
import matplotlib.pyplot as plt

crottes = df[(df["Espèce_Fact_Nat2"] == "Eléphant de forêt") & (df["Type indice "] == "Crotte")].groupby("Transec N°").size().reindex(L_km.index, fill_value=0) / L_km
strate = df.drop_duplicates("Transec N°").set_index("Transec N°")["Strate/site"].reindex(L_km.index)
couleurs = {"Parc national": "#2e7651", "Zone tampon": "#c9a227", "Concession forestière": "#b2433a"}

fig, ax = plt.subplots(figsize=(7, 4.5))
for s, c in couleurs.items():
    m = strate == s
    ax.scatter(pression[m], crottes[m], label=s, color=c)
ax.set_xlabel("Indices humains par km")
ax.set_ylabel("Crottes d'éléphant par km")
ax.legend()
plt.show()
print("Corrélation de Spearman :", round(crottes.corr(pression, method="spearman"), 2))
```
