### 5a · Le concept : l'index

Chaque tableau et chaque Series a un **index** : les étiquettes des lignes (affichées à gauche). Par défaut ce sont des numéros. `.set_index("colonne")` utilise une colonne comme étiquettes. Intérêt : quand on divise deux Series, pandas **associe les valeurs qui ont la même étiquette**, quel que soit leur ordre.

```python
a = pd.Series({"T1": 4, "T2": 6})
b = pd.Series({"T2": 3, "T1": 2})
print(a / b)  # T1 : 4/2, T2 : 6/3, malgré l'ordre différent
```

### 9a · Le concept : le hasard reproductible

`rng = np.random.default_rng(graine)` crée un générateur de nombres aléatoires. Avec la même graine, on obtient toujours les mêmes tirages : indispensable pour qu'un résultat soit vérifiable. `rng.choice(liste, size=n, replace=True)` tire n éléments **avec remise** : un même élément peut sortir plusieurs fois.

```python
rng = np.random.default_rng(7)
print(rng.choice(["a", "b", "c"], size=6, replace=True))
```
