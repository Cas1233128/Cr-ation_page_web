### 10a · Le concept : comparer

### 17a · Le concept : les fonctions

Une **fonction** est un calcul réutilisable. `def` la définit, les **paramètres** sont entre parenthèses, `return` renvoie le résultat.

> `print` affiche une valeur à l'écran. `return` la **renvoie** pour qu'on puisse la réutiliser. Ce n'est pas la même chose.

```python
def en_hectares(m2):
    return m2 / 10000

resultat = en_hectares(5000)
print(resultat)
print(en_hectares(25000))
```
