### 6a · Le concept : transform

`groupby(...).mean()` renvoie **une ligne par groupe**. `groupby(...).transform("mean")` renvoie **une valeur par ligne du tableau** : la moyenne de son groupe. C'est ce qu'il faut pour comparer chaque arbre à son groupe, ou pour remplir des trous.

```python
print(arbres.groupby("placette")["dhp"].mean())
print(arbres.groupby("placette")["dhp"].transform("mean").tolist())
```

### 8a · Le concept : un graphique avec matplotlib

On crée une figure et des axes avec `fig, ax = plt.subplots()`, puis on dessine sur `ax` : `ax.hist` (histogramme), `ax.bar` / `ax.barh` (barres), `ax.scatter` (nuage de points), `ax.plot` (courbe). On termine par les légendes d'axes et `plt.show()`.

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot([1, 2, 3], [10, 30, 20], marker="o")
ax.set_xlabel("Campagne")
ax.set_ylabel("Nids observés")
plt.show()
```
